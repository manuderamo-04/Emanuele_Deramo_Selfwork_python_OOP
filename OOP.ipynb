```python
from abc import ABC, abstractmethod

class Persona(ABC):
    def __init__(self, nome, cognome, eta):
        self.nome = nome
        self.cognome = cognome
        self.eta = eta
        
    @abstractmethod
    def get_ruolo(self):
        """Metodo astratto che deve essere implementato dalle sottoclassi"""
        pass

class Studente(Persona):
    def __init__(self, nome, cognome, eta, matricola, corso_di_studi):
        super().__init__(nome, cognome, eta)
        self.matricola = matricola
        self.corso_di_studi = corso_di_studi
        
    def get_ruolo(self):
        return f"Studente iscritto a {self.corso_di_studi} (Matricola: {self.matricola})"

class Dipendente(Persona):
    def __init__(self, nome, cognome, eta, stipendio_base, codice_dipendente):
        super().__init__(nome, cognome, eta)
        self.stipendio_base = stipendio_base
        self.codice_dipendente = codice_dipendente
        
    def get_ruolo(self):
        return f"Dipendente aziendale (Codice: {self.codice_dipendente})"

class Docente(Dipendente):
    def __init__(self, nome, cognome, eta, stipendio_base, codice_dipendente, materia, ore_settimanali):
        super().__init__(nome, cognome, eta, stipendio_base, codice_dipendente)
        self.materia = materia
        self.ore_settimanali = ore_settimanali
        
    def get_ruolo(self):
        return f"Docente di {self.materia} ({self.ore_settimanali} ore settimanali)"

class Freelance(Dipendente):
    def __init__(self, nome, cognome, eta, stipendio_base, codice_dipendente, partita_iva, tariffa_oraria):
        super().__init__(nome, cognome, eta, stipendio_base, codice_dipendente)
        self.partita_iva = partita_iva
        self.tariffa_oraria = tariffa_oraria
        
    def get_ruolo(self):
        return f"Consulente Freelance (P.IVA: {self.partita_iva} - Tariffa: {self.tariffa_oraria}€/h)"


if __name__ == "__main__":

    studente1 = Studente("Mario", "Rossi", 22, "M63001234", "Informatica")
    
    docente1 = Docente("Laura", "Bianchi", 45, 1800, "DIP-998", "Programmazione Python", 18)
    
    freelance1 = Freelance("Luca", "Verdi", 30, 2200, "DIP-452", "IT12345678901", 45)


    print(studente1.get_ruolo())
    print(docente1.get_ruolo())
    print(freelance1.get_ruolo())
```
